```python
import numpy as np
```

```python
def f(x):
    """ Defined function to apply Newton Raphson method to """
    return x*x + 2*x + 1
    
def d(x, f):
    """ Derivative funtion """
    eps = 0.00001
    a = (f(x+eps)-f(x))
    return a/eps

def dd(x, f):
    """ Second Derivative funtion """
    eps = 0.00000001
    a = (d(x+eps, f)-d(x, f))
    return a/eps

def neuton_raphson(x_0, f, eps):
    """ Newton Raphson method """
    x_t_1 = x_0
    a = d(x_t_1, f)
    b = dd(x_t_1, f)
    x_t = x_t_1 - a/b
    while abs(x_t - x_t_1)>eps:
        
        a = d(x_t_1, f)
        b = dd(x_t_1, f)
        temp = x_t_1 - a/b
        x_t = temp - d(temp, f)/dd(temp, f)
        x_t_1 = temp
        
    return x_t
```

```python
neuton_raphson(10, np.cos, 0.00001)
```
