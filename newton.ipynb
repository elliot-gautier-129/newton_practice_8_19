```python
import numpy as np
from scipy import differentiate
```

```python
def f_mult(point):
    """ New function where x is array and f(x) is univariate """
    x, y = point[0], point[1]
    return 2*x*x + y + x + 3
    

def mult_newton_raphson(x_0, f, thresh = 0.001):
    """ Multivariate Newton Raphson method """
    
    x_t_1 = x_0
    H = differentiate.hessian(f, x_t_1)
    J = differentiate.jacobian(f, x_t_1)
    x_t = x_t_1 - np.linalg.solve(H['status'],J['df'])
    
    while np.linalg.norm(x_t - x_t_1)>thresh:

        H = differentiate.hessian(f, x_t_1)
        J = differentiate.jacobian(f, x_t_1)
        temp = x_t_1 - np.linalg.solve(H['status'],J['df'])

        H_1 = differentiate.hessian(f, temp)
        J_1 = differentiate.jacobian(f, temp)
        x_t = temp - np.linalg.solve(H_1['status'], J_1['df'])
        x_t_1 = temp
        
    return x_t
```

```python
import numpy as np
import differentiate  # Assuming your auto-diff module


def mult_newton_raphson(x_0, f, thresh=0.001, max_iter=100):
    """Multivariate Newton-Raphson method for optimization/finding critical points."""
    x_curr = np.array(x_0, dtype=float)

    for _ in range(max_iter):
        # 1. Compute Hessian and Jacobian
        H_res = differentiate.hessian(f, x_curr)
        J_res = differentiate.jacobian(f, x_curr)

        # Ensure correct matrix extraction (adjust keys based on your specific library)
        H = H_res["hessian"] if "hessian" in H_res else H_res["status"]
        J = J_res["df"]

        # 2. Check for singular Hessian or invertibility issues
        if np.abs(np.linalg.det(H)) < 1e-12:
            raise ValueError(
                "Hessian is singular or nearly singular. Newton-Raphson cannot proceed."
            )

        # 3. Compute Newton step: delta = H^(-1) * J
        delta = np.linalg.solve(H, J)

        # 4. Update position
        x_next = x_curr - delta

        # 5. Convergence check
        if np.linalg.norm(delta) < thresh:
            return x_next

        x_curr = x_next

    return x_curr
```
