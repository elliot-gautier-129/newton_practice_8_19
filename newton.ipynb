```python
def f(x):
    return x*x + 2*x + 1
    
def d(x):
    eps = 0.00001
    a = (f(x+eps)-f(x))
    return a/eps

def dd(x): 
    eps = 0.00000001
    a = (d(x+eps)-d(x))
    return a/eps

def neuton_raphson(x_0, eps):
    x_t_1 = x_0
    a = d(x_t_1)
    b = dd(x_t_1)
    x_t = x_t_1 - a/b
    while abs(x_t - x_t_1)>eps:
        a = d(x_t_1)
        b = dd(x_t_1)
        temp = x_t_1 - a/b
        x_t = temp - d(temp)/dd(temp)
        x_t_1 = temp
        
    return x_t
```
